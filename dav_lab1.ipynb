```python
import pandas as pd
data={
    "names":['a','b','c','d','e','f','g'],
    "branch":['cse','aiml','eee','ds','ip','me','ise'],
    "marks":[50,60,70,90,80,40,30]
}
df=pd.DataFrame(data)
print(df)
df.head()
```

```python
df.describe()
```

```python
df.tail()
```

```python
print("marks column")
print(df['marks'])
```

```python
print("\nbranch and marks:")
print(df[["branch","marks"]])
```

```python
print("first row: ")
print(df.iloc[0])            #integer location based indxing

print("\nsecond row:")
print(df.iloc[1])

print("\nfirst three rows:")
print(df.iloc[0:3])
print("\nfirst two rows and columns:")
print(df.iloc[0:2, 0:2])
```

```python
high_scores=df[df["marks"]>45]
print(high_scores)
```

```python
print("students from cse:")
display(df[df["branch"]=="cse"])

print("students scoring above 85:")
display(df[df["marks"]>40])
```
